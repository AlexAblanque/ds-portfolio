## 1. Perform Standard Imports


```python
import numpy as np
import torch

print("NumPy version:", np.__version__)
print("PyTorch version:", torch.__version__)
```

## 2. Create `set_seed(seed: int)`

The function sets the random seed for both NumPy and PyTorch and does not return a value.


```python
def set_seed(seed: int) -> None:
    np.random.seed(seed)
    torch.manual_seed(seed)
```

## 3. Create NumPy array `arr`

Create 6 random integers from **0 (inclusive) to 5 (exclusive)** using seed `42`.

`dtype=np.int32` is used so that Task 5 can explicitly change the tensor from `int32` to `int64`.


```python
set_seed(42)

arr = np.random.randint(0, 5, size=6).astype(np.int32)

print("arr =", arr)
print("arr dtype =", arr.dtype)
```

## 4. Create tensor `x` from `arr`


```python
x = torch.from_numpy(arr)

print("x =", x)
print("x dtype =", x.dtype)
```

## 5. Change `x` from `int32` to `int64`


```python
x = x.type(torch.int64)

print("x =", x)
print("New dtype =", x.dtype)
```

## 6. Reshape `x` into a 3 × 2 tensor


```python
x = x.reshape(3, 2)

print("x =")
print(x)
print("Shape =", x.shape)
```

## 7. Return the right-hand column of `x`


```python
right_column = x[:, 1]

print("Right-hand column:")
print(right_column)
```

## 8. Return the square values of `x` without changing `x`

`x ** 2` creates a new tensor, so the original `x` remains unchanged.


```python
x_squared = x ** 2

print("Squared values:")
print(x_squared)

print("\nOriginal x is unchanged:")
print(x)
```

## 9. Create tensor `y`

Since `x` has shape **(3, 2)**, `y` must have **2 rows** for matrix multiplication.

To give `y` the same number of elements as `x` (6 elements), its shape is **(2, 3)**.


```python
torch.manual_seed(42)

y = torch.randint(0, 5, (2, 3))

print("y =")
print(y)
print("Shape =", y.shape)
print("Number of elements =", y.numel())
```

## 10. Find the matrix product of `x` and `y`

`x` has shape **(3, 2)** and `y` has shape **(2, 3)**, so the resulting matrix has shape **(3, 3)**.


```python
matrix_product = x @ y

print("x =")
print(x)

print("\ny =")
print(y)

print("\nx @ y =")
print(matrix_product)

print("\nProduct shape =", matrix_product.shape)
```
