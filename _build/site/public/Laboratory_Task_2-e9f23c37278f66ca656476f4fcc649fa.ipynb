# Laboratory Task 2 — Single Forward Pass and Error

**Instruction:** Perform a single forward pass and compute for the error.

Given $x=[1,0,1]^T$, target $y=1$, and ReLU $f(z)=\max(0,z)$.

## Given parameters

Hidden weights: $w_{11}=0.2$, $w_{12}=-0.3$, $w_{13}=0.4$, $w_{14}=0.1$, $w_{15}=-0.5$, $w_{16}=0.2$.

Output weights: $w_{21}=-0.3$, $w_{22}=-0.2$.

Biases: $\theta_1=-0.4$, $\theta_2=0.2$, $\theta_3=0.1$.

## Step 1 — Hidden layer

$Z_1=(0.2)(1)+(0.4)(0)+(-0.5)(1)-0.4=-0.7$

$H_1=\max(0,-0.7)=0$

$Z_2=(-0.3)(1)+(0.1)(0)+(0.2)(1)+0.2=0.1$

$H_2=\max(0,0.1)=0.1$

## Step 2 — Output layer

$Z_3=(-0.3)(0)+(-0.2)(0.1)+0.1=0.08$

$\hat{y}=\max(0,0.08)=0.08$

## Step 3 — Error

Using the error convention from the lesson, $\delta=\hat{y}-y$:

$\delta=0.08-1=\mathbf{-0.92}$

For reference, squared error is $(1-0.08)^2=\mathbf{0.8464}$, and half-squared error is $0.4232$.
