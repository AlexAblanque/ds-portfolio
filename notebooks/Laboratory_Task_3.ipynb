# Laboratory Task 3 — Forward and Backward Propagation

**Instruction:** Perform forward and backward propagation in Python using the inputs from Laboratory Task 2. ReLU is used as the activation function and the learning rate is **0.001**.

```python
import numpy as np
np.set_printoptions(precision=6, suppress=True)
x=np.array([1.,0.,1.]); y=np.array([1.])
lr=0.001
W_h=np.array([[0.2,-0.3],[0.4,0.1],[-0.5,0.2]],dtype=float)
b_h=np.array([-0.4,0.2],dtype=float)
W_o=np.array([[-0.3],[-0.2]],dtype=float)
b_o=np.array([0.1],dtype=float)
def relu(z): return np.maximum(0,z)
def relu_prime(z): return (z>0).astype(float)
```

## 1. Forward propagation

```python
z_h=x@W_h+b_h
a_h=relu(z_h)
z_o=a_h@W_o+b_o
y_hat=relu(z_o)
error=y_hat-y
loss=0.5*np.sum(error**2)
print('z_h =',z_h); print('a_h =',a_h); print('z_o =',z_o); print('y_hat =',y_hat); print('error =',error); print('half-squared loss =',loss)
```

## 2. Backward propagation

For $E=\frac12(\hat y-y)^2$, the output delta is $(\hat y-y)f'(Z_3)$. The error is then propagated to the hidden layer using the output weights and the derivative of ReLU.

```python
delta_o=error*relu_prime(z_o)
grad_W_o=np.outer(a_h,delta_o)
grad_b_o=delta_o.copy()
delta_h=(W_o.flatten()*delta_o.item())*relu_prime(z_h)
grad_W_h=np.outer(x,delta_h)
grad_b_h=delta_h.copy()
print('Output delta:',delta_o); print('Hidden delta:',delta_h); print('dE/dW_o:\n',grad_W_o); print('dE/db_o:',grad_b_o); print('dE/dW_h:\n',grad_W_h); print('dE/db_h:',grad_b_h)
```

## 3. Update weights and biases

```python
W_h_new=W_h-lr*grad_W_h; b_h_new=b_h-lr*grad_b_h
W_o_new=W_o-lr*grad_W_o; b_o_new=b_o-lr*grad_b_o
print('Updated hidden weights:\n',W_h_new); print('Updated hidden biases:',b_h_new); print('Updated output weights:\n',W_o_new); print('Updated output bias:',b_o_new)
```

## 4. Check the result after one update

```python
z_h2=x@W_h_new+b_h_new; a_h2=relu(z_h2); z_o2=a_h2@W_o_new+b_o_new; y_hat2=relu(z_o2); loss2=0.5*np.sum((y_hat2-y)**2)
print('New prediction:',y_hat2); print('New half-squared loss:',loss2); print('Loss decreased:',loss2 < loss)
```
